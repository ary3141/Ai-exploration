## Load Data

```python
# LOAD COCO

def load_coco_annotations(annotation_file):
    """
    Load COCO annotations, remove unused categories, and return a dictionary of valid categories
    with zero-based indexing.
    """
    coco = COCO(annotation_file)

    # Ambil semua category ID yang digunakan dalam gambar
    used_category_ids = set()
    for img_id in coco.getImgIds():
        ann_ids = coco.getAnnIds(imgIds=img_id)
        anns = coco.loadAnns(ann_ids)
        used_category_ids.update(ann['category_id'] for ann in anns)

    # Filter hanya kategori yang digunakan dan urutkan
    valid_category_ids = sorted(used_category_ids)  # Urutkan ID yang masih digunakan

    # Buat dictionary dengan format {category_name: index}
    category_dict = {coco.loadCats(cat_id)[0]['name']: idx for idx, cat_id in enumerate(valid_category_ids)}

    return coco, category_dict
```

## Preprocess

```python
# CREAT ARRAY MASKING

def coco_to_multilayer_mask(coco, image_id, image_size, category_dict):
    height, width = image_size
    num_classes = len(category_dict)

    # Ubah mask ke format (H, W, num_classes)
    mask = np.zeros((height, width, num_classes), dtype=np.uint8)

    ann_ids = coco.getAnnIds(imgIds=[image_id])
    anns = coco.loadAnns(ann_ids)

    for ann in anns:
        category_id = ann["category_id"]
        category_name = coco.loadCats([category_id])[0]["name"]  # Tetap [0]

        if category_name in category_dict:
            layer_index = category_dict[category_name]
            segmentation = coco.annToMask(ann)
            mask[:, :, layer_index] = np.maximum(mask[:, :, layer_index], segmentation)  # Perubahan utama

    return mask
```

```python
# CONVERT ARRAY MASK TO TIFF

def save_multilayer_tiff(array, filename):
    if array.ndim != 3:
        raise ValueError("Input array must be 3D with shape (H, W, C)")

    height, width, num_layers = array.shape  # Format tetap (H, W, C)

    transform = from_origin(0, 0, 1, 1)  # Dummy transform for non-geospatial data

    metadata = {
        "driver": "GTiff",
        "height": height,
        "width": width,
        "count": num_layers,
        "dtype": array.dtype,
        "crs": None,
        "transform": transform
    }

    with rasterio.open(filename, "w", **metadata) as dst:
        for i in range(num_layers):
            dst.write(array[:, :, i], i + 1)  # Tetap menggunakan (H, W, C)
```

```python
# CREATE FILE TIFF

def process_dataset(image_folder, annotation_file, output_folder):
    os.makedirs(output_folder, exist_ok=True)

    # Load COCO annotations dan category_dict
    coco, category_dict = load_coco_annotations(annotation_file)

    for img_info in coco.dataset["images"]:
        image_id = img_info["id"]
        file_name = img_info["file_name"]
        image_size = (img_info["height"], img_info["width"])

        # Generate multi-layer mask menggunakan category_dict
        mask = coco_to_multilayer_mask(coco, image_id, image_size, category_dict)
        output_path = os.path.join(output_folder, file_name.replace(".jpg", ".tiff"))

        # Simpan hasil mask sebagai file TIFF
        save_multilayer_tiff(mask, output_path)
        # print(f"Saved: {output_path}")
```

```python
import random
import os
import rasterio
import matplotlib.pyplot as plt

def display_random_tiff(directory, category_dict):
    """
    Display a randomly selected TIFF file from a directory.
    Each layer is shown in a separate subplot with category information.

    Parameters:
        directory (str): Path to the directory containing TIFF files.
        category_dict (dict): Dictionary mapping layer index to category name.
    """
    # Ambil semua file TIFF dalam folder
    tiff_files = [f for f in os.listdir(directory) if f.endswith(".tiff")]

    if not tiff_files:
        print("No TIFF files found in the directory.")
        return

    # Pilih satu file secara acak
    random_file = random.choice(tiff_files)
    file_path = os.path.join(directory, random_file)

    # Buka file TIFF dengan rasterio
    with rasterio.open(file_path) as src:
        num_layers = src.count  # Jumlah kategori dalam TIFF
        masks = np.stack([src.read(i + 1) for i in range(num_layers)], axis=-1)  # Ubah ke (H, W, C)

    height, width, num_layers = masks.shape  # Dapatkan ukuran

    # Tampilkan semua layer dalam satu baris
    fig, axes = plt.subplots(1, num_layers, figsize=(num_layers * 3, 3))

    if num_layers == 1:
        category_name = category_dict.get(0, "Unknown")
        axes.imshow(masks[:, :, 0], cmap="gray")  # Ambil layer pertama dengan format (H, W)
        axes.set_title(f"Layer 1: {category_name}")
        axes.axis("off")
    else:
        for i, ax in enumerate(axes):
            category_name = category_dict.get(i, "Unknown")  # Ambil nama kategori
            ax.imshow(masks[:, :, i], cmap="gray")  # Ambil setiap layer dari (H, W, C)
            ax.set_title(f"Layer {i + 1}: {category_name}")
            ax.axis("off")

    plt.suptitle(f"Displaying: {random_file}")
    plt.show()
```

## Data Splitting

```python
# Path ke folder dataset
image_folder = "../content/Scan-Gizi-Makanan-3/train/"
mask_folder = "../content/Scan-Gizi-Makanan-3/train/mask/"
target_size = (256, 256)
num_classes = len(dictclass)
```

```python
image_files = sorted([f for f in os.listdir(image_folder) if f.endswith(('.png', '.jpg', '.jpeg'))])
mask_files = sorted([f for f in os.listdir(mask_folder) if f.endswith('.tiff')])
```

```python
image_mask_pairs = []
for img_name in image_files:
    mask_name = img_name.rsplit('.', 1)[0] + ".tiff"  # Ubah ekstensi gambar ke mask TIFF
    if mask_name in mask_files:
        image_mask_pairs.append((os.path.join(image_folder, img_name), os.path.join(mask_folder, mask_name)))



train_pairs, test_pairs = train_test_split(image_mask_pairs, train_size=0.8, random_state=42)

train_image_paths, train_mask_paths = zip(*train_pairs)
test_image_paths, test_mask_paths = zip(*test_pairs)
```

```python
train_image_paths = np.array(train_image_paths, dtype=str)
train_mask_paths = np.array(train_mask_paths, dtype=str)
test_image_paths = np.array(test_image_paths, dtype=str)
test_mask_paths = np.array(test_mask_paths, dtype=str)
```

```python
def load_image(image_path):
    image = cv2.imread(image_path.decode())  # Baca gambar (BGR)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert ke RGB
    image = cv2.resize(image, target_size)  # Resize
    image = image / 255.0  # Normalisasi (0-1)
    return image.astype(np.float32)

def load_mask(mask_path):
    mask = tiff.imread(mask_path.decode())  # Baca TIFF
    mask = cv2.resize(mask, target_size)  # Resize
    mask = mask.astype(np.float32)  # Pastikan tipe float32
    if mask.shape[-1] != num_classes:
        raise ValueError(f"Mask shape {mask.shape} tidak sesuai dengan num_classes={num_classes}")
    return mask
```

```python
def preprocess(image_path, mask_path):
    image = tf.numpy_function(load_image, [image_path], tf.float32)
    mask = tf.numpy_function(load_mask, [mask_path], tf.float32)

    image.set_shape((256, 256, 3))
    mask.set_shape((256, 256, num_classes))

    return image, mask
```

```python
train_dataset = tf.data.Dataset.from_tensor_slices((train_image_paths, train_mask_paths))
test_dataset = tf.data.Dataset.from_tensor_slices((test_image_paths, test_mask_paths))

train_dataset = train_dataset.map(preprocess).shuffle(100).batch(16).prefetch(tf.data.AUTOTUNE)
test_dataset = test_dataset.map(preprocess).batch(16).prefetch(tf.data.AUTOTUNE)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import EfficientNetV2S
```

```python
def build_encoder(input_shape):
    inputs = layers.Input(shape=input_shape)

    base_model = EfficientNetV2S(include_top=False, weights=None, input_tensor=inputs)

    skip_layers = [
        "block2b_add",  # (128, 128, C)
        "block3b_add",  # (64, 64, C)
        "block4b_add",  # (32, 32, C)
        "block6b_add",  # (16, 16, C)
    ]

    skips = [base_model.get_layer(layer).output for layer in skip_layers]

    bottleneck = base_model.output  # (8, 8, C)

    return Model(inputs, [bottleneck] + skips)
```

```python
def build_decoder(bottleneck, skips, num_classes):

    x = bottleneck

    # Decoder Block 1: Upsampling dari bottleneck (8x8 → 16x16)
    x = layers.Conv2DTranspose(512, (3, 3), strides=2, padding="same", activation="relu")(x)  # (16, 16, 512)
    skips[3] = layers.UpSampling2D((2, 2))(skips[3])
    x = layers.concatenate([x, skips[3]])

    # Decoder Block 2: 16x16 → 32x32
    x = layers.Conv2DTranspose(256, (3, 3), strides=2, padding="same", activation="relu")(x)  # (32, 32, 256)
    skips[2] = layers.UpSampling2D((2, 2))(skips[2])
    x = layers.concatenate([x, skips[2]])

    # Decoder Block 3: 32x32 → 64x64
    x = layers.Conv2DTranspose(128, (3, 3), strides=2, padding="same", activation="relu")(x)  # (64, 64, 128)
    skips[1] = layers.UpSampling2D((2, 2))(skips[1])
    x = layers.concatenate([x, skips[1]])

    # Decoder Block 4: 64x64 → 128x128
    x = layers.Conv2DTranspose(64, (3, 3), strides=2, padding="same", activation="relu")(x)  # (128, 128, 64)
    skips[0] = layers.UpSampling2D((2, 2))(skips[0])
    x = layers.concatenate([x, skips[0]])

    # Decoder Block 5: 128x128 → 256x256 (Final Output)
    x = layers.Conv2DTranspose(32, (3, 3), strides=2, padding="same", activation="relu")(x)  # (256, 256, 32)

    # Output Layer
    outputs = layers.Conv2D(num_classes, (1, 1), activation="sigmoid")(x)  # (256, 256, num_classes)

    return outputs
```

```python
def build_eff_unet(input_shape=(256, 256, 3), num_classes=10):
    # Encoder
    encoder = build_encoder(input_shape)
    outputs = encoder.output

    bottleneck = outputs[0]
    skips = outputs[1:]

    # Decoder
    outputs = build_decoder(bottleneck, skips, num_classes)

    model = Model(encoder.input, outputs, name="Eff-UNet")
    return model
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import EfficientNetB7

# 🟢 DICE Coefficient
def dice_coefficient(y_true, y_pred, smooth=1e-6):
    y_true_f = tf.keras.backend.flatten(y_true)
    y_pred_f = tf.keras.backend.flatten(y_pred)
    intersection = tf.keras.backend.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (tf.keras.backend.sum(y_true_f) + tf.keras.backend.sum(y_pred_f) + smooth)

# 🟢 Dice Loss
def dice_loss(y_true, y_pred):
    return 1 - dice_coefficient(y_true, y_pred)

def conv2d_block(input_tensor, n_filters, kernel_size=3):
    x = input_tensor
    for _ in range(2):
        x = layers.Conv2D(filters=n_filters, kernel_size=(kernel_size, kernel_size),
                          kernel_initializer='he_normal', padding='same')(x)
        x = layers.Activation('relu')(x)
    return x

def encoder_block(input_tensor):
    base_model = EfficientNetB7(include_top=False, weights=None, input_tensor=input_tensor)

    # 🟢 Ambil 6 skip connections untuk stabilitas model
    skips = [
        base_model.get_layer("block1a_activation").output,
        base_model.get_layer("block2a_expand_activation").output,
        base_model.get_layer("block3a_expand_activation").output,
        base_model.get_layer("block4a_expand_activation").output,
        base_model.get_layer("block5a_expand_activation").output,
        base_model.get_layer("block6a_expand_activation").output,
    ]

    bottleneck = base_model.get_layer("top_activation").output  # (4,4,1536) → Bottleneck

    return base_model.input, skips, bottleneck

def decoder_block(inputs, skip_tensor, n_filters, kernel_size=3):
    x = layers.Conv2DTranspose(n_filters, kernel_size, strides=(2, 2), padding='same')(inputs)

    if skip_tensor is not None:
        x_shape = tf.keras.backend.int_shape(x)[1:3]
        skip_shape = tf.keras.backend.int_shape(skip_tensor)[1:3]

        # Pastikan ukuran skip_tensor sama dengan x sebelum Concatenate
        if x_shape != skip_shape:
            x = layers.Resizing(skip_shape[0], skip_shape[1])(x)  # Resize pakai layer Resizing

        x = layers.Concatenate()([x, skip_tensor])  # Gabungkan dengan skip connection

    x = conv2d_block(x, n_filters)
    return x

def build_eff_unet(input_shape=(256, 256, 3), num_classes=10):
    inputs = layers.Input(shape=input_shape)  # 🟢 Buat input di sini
    skips, bottleneck = encoder_block(inputs)  # 🔄 Gunakan `inputs`

    bottleneck = conv2d_block(bottleneck, 1024)  # Tambahkan conv setelah bottleneck

    d1 = decoder_block(bottleneck, skips[5], 512)
    d2 = decoder_block(d1, skips[4], 256)
    d3 = decoder_block(d2, skips[3], 128)
    d4 = decoder_block(d3, skips[2], 96)
    d5 = decoder_block(d4, skips[1], 64)
    d6 = decoder_block(d5, skips[0], 32)

    outputs = layers.Conv2D(num_classes, (1, 1), activation="sigmoid")(d6)
    model = Model(inputs, outputs, name="EfficientUNet")

    model.compile(optimizer=tf.keras.optimizers.Adam(), loss=dice_loss, metrics=[dice_coefficient])
    return model
```
